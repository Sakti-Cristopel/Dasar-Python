```python
url = "https://drive.google.com/uc?id=1LfQWProB0VjWN5q8bKuRIgn-stULfIRo"

df = pd.read_csv(url)
```

```python
print("=== INFO DATASET ===")
df.info()

print("\n=== DESKRIPSI DATA ===")
print(df.describe())

print("\n=== MISSING VALUES ===")
print(df.isnull().sum())

print("\nShape Awal :", df.shape)
```

```python
df.drop_duplicates(inplace=True)

print("Jumlah data setelah hapus duplikat:", df.shape)
```

```python
df['kota'] = df['kota'].str.strip().str.title()
df['kondisi'] = df['kondisi'].str.strip().str.lower()
```

```python
df['luas_m2'] = df['luas_m2'].fillna(df['luas_m2'].median())

df['harga_juta'] = df['harga_juta'].fillna(
    df['harga_juta'].median()
)

df['kamar'] = df['kamar'].fillna(
    df['kamar'].mode()[0]
)
```

```python
for col in ['harga_juta', 'luas_m2', 'tahun_bangun']:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[col] = df[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )
```

```python
print("\n=== VALIDASI ===")

print("Total Missing Value:")
print(df.isnull().sum())

print("\nJumlah Missing Keseluruhan:")
print(df.isnull().sum().sum())

print("\nJumlah Duplikat:")
print(df.duplicated().sum())
```

```python
df.to_csv('housing_clean.csv', index=False)

print("Dataset bersih berhasil disimpan!")
```
