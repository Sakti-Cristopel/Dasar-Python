```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
```

```python
df = sns.load_dataset('tips')

print("Shape Dataset:")
print(df.shape)

print("\nTipe Data:")
print(df.dtypes)

print("\nStatistik Deskriptif:")
print(df.describe().round(2))
```

```python
fig = plt.figure(figsize=(16,12))

fig.suptitle(
    'Dashboard Analisis Dataset Tips',
    fontsize=18,
    fontweight='bold'
)

gs = gridspec.GridSpec(
    2,
    2,
    figure=fig,
    hspace=0.4,
    wspace=0.3
)

ax1 = fig.add_subplot(gs[0,0])
ax2 = fig.add_subplot(gs[0,1])
ax3 = fig.add_subplot(gs[1,0])
ax4 = fig.add_subplot(gs[1,1])

# Grafik 1 - Bar Chart
avg_day = df.groupby('day')['total_bill'].mean()

avg_day.plot(
    kind='bar',
    color='steelblue',
    ax=ax1
)

ax1.set_title('Rata-rata Total Bill per Hari')
ax1.set_xlabel('Hari')
ax1.set_ylabel('Total Bill')

# Grafik 2 - Histogram
sns.histplot(
    data=df,
    x='total_bill',
    kde=True,
    color='green',
    ax=ax2
)

ax2.set_title('Distribusi Total Bill')

# Grafik 3 - Boxplot
sns.boxplot(
    data=df,
    x='day',
    y='total_bill',
    ax=ax3
)

ax3.set_title('Boxplot Total Bill per Hari')

# Grafik 4 - Scatter Plot
sns.scatterplot(
    data=df,
    x='total_bill',
    y='tip',
    hue='time',
    ax=ax4
)

ax4.set_title('Hubungan Total Bill dan Tip')

plt.tight_layout()

plt.show()
```

```python
plt.savefig(
    'dashboard_tips.png',
    dpi=150,
    bbox_inches='tight'
)

print("Dashboard berhasil disimpan.")
```
